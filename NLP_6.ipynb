```python
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
nltk.download('punkt')
nltk.download('stopwords')
import nltk
nltk.download('punkt_tab')
```

```python
data = {
    "Category":[
        "Sports","Sports","Sports","Sports",
        "Politics","Politics","Politics","Politics",
        "Technology","Technology","Technology","Technology",
        "Entertainment","Entertainment","Entertainment","Entertainment",
        "Business","Business","Business","Business"
    ],

    "Article":[

        "India defeated Australia in the cricket world cup match.",
        "Lionel Messi scored two goals in the football tournament.",
        "Virat Kohli played an outstanding innings yesterday.",
        "The Olympic games will begin next month.",

        "The parliament passed a new education bill.",
        "The prime minister addressed the nation.",
        "Election campaigns have started across the country.",
        "The government announced a new tax policy.",

        "Artificial Intelligence is changing healthcare.",
        "Google launched a new AI-powered search feature.",
        "Python is widely used for machine learning.",
        "Cybersecurity threats are increasing every year.",

        "The new movie broke all box office records.",
        "The actor received the best performance award.",
        "Fans celebrated the music concert.",
        "The film trailer gained millions of views.",

        "Stock markets closed at a record high.",
        "The company reported huge quarterly profits.",
        "Investors are buying technology shares.",
        "The startup raised funding from investors."
    ]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
stop_words = set(stopwords.words("english"))

def preprocess(text):

    text = text.lower()

    text = re.sub(r'[^a-zA-Z ]', '', text)

    words = word_tokenize(text)

    words = [word for word in words if word not in stop_words]

    return " ".join(words)

df["Clean_Text"] = df["Article"].apply(preprocess)

print(df.head())
```

```python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["Clean_Text"])

y = df["Category"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
```

```python
model = MultinomialNB()

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, predictions))

print("\nClassification Report")

print(classification_report(y_test, predictions))
```

```python
print(confusion_matrix(y_test, predictions))
```

```python
news = [
    "Apple launched a new AI powered smartphone with advanced processor."
]

news_clean = [preprocess(x) for x in news]

news_vector = vectorizer.transform(news_clean)

prediction = model.predict(news_vector)

print("Category:", prediction[0])
```
